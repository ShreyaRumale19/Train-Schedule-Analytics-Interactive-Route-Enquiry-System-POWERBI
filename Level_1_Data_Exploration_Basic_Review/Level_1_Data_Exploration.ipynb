```python
total_records = df.shape[0]
total_attributes = df.shape[1]

print("Total Records:", total_records)
print("Total Attributes:", total_attributes)
```

```python
total_trains = df["Train_No"].nunique()

print("Total Unique Trains:", total_trains)
```

```python
df["Train_No"].unique()
```

```python
stops_per_train = df.groupby("Train_No")["Station_Code"].count()

stops_per_train.head()
```

```python
#Maximum and Minimum Stops
max_stops = stops_per_train.max()
min_stops = stops_per_train.min()

print("Maximum stops:", max_stops)
print("Minimum stops:", min_stops)
```

```python
max_stop_trains = stops_per_train[
    stops_per_train == max_stops
]

min_stop_trains = stops_per_train[
    stops_per_train == min_stops
]

print("Train(s) with maximum stops:")
print(max_stop_trains)

print("\nTrain(s) with minimum stops:")
print(min_stop_trains)
```

##  Level 1 Findings

- The dataset contains 186,074 records and 12 attributes.
- The dataset contains 11,113 unique trains.
- Each train is represented through its station-wise schedule records.
- Train 53041 has the maximum number of scheduled station records, with 118 stops.
- The minimum number of scheduled station records is 2.
- A total of 1,249 trains have the minimum of 2 scheduled station records.
- The analysis provides a basic understanding of train coverage, station-level records, and the variation in the number of stops across trains.
